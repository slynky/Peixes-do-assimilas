# Criar os animais pro mateus

````python
from google import genai
from google.genai import types
from config import chave
import pandas as pd
import io
import time

client = genai.Client(api_key=chave)

prompt_text = f"""
Crie uma lista de 15 animais terrestres e 5 aquaticos com base nos dados fornecidos.
Contexto: RPG de mesa de sobrevivência. Apenas animais médios e grandes selvagens que representem um desafio (esquivar, lutar ou fugir).
Níveis de perigo: quase inofensivo, ofensivo, matador.

Retorne APENAS os dados em formato CSV, separados por ponto e vírgula (;).
NÃO inclua introduções ou conclusões.
NÃO use blocos de código markdown (sem ```), apenas o texto cru.

O Cabeçalho deve ser exatamente: Nome;Tamanho;Horario;Alimentacao;Agressividade;Caracteristicas
"""

try:
    print("Gerando conteúdo...")
    response = client.models.generate_content(
        model='gemini-2.5-flash',
        contents=prompt_text,
        config=types.GenerateContentConfig(
            temperature=0.7
        )
    )

    dados_limpos = response.text.replace("```csv", "").replace("```", "").strip()

    arquivo_virtual = io.StringIO(dados_limpos)

    df = pd.read_csv(arquivo_virtual, sep=';')

    nome_arquivo = 'animais.xlsx'
    df.to_excel(nome_arquivo, index=False, engine='openpyxl')

    print(f"Sucesso! Arquivo '{nome_arquivo}' criado corretamente.")
    time.sleep(2)

except Exception as e:
    print(f"Erro ao processar: {e}")
    if 'response' in locals():
        print("Conteúdo retornado pela IA:", response.text)

print("\nProcesso finalizado.")
````
