# Criar os peixes da campanha do assimilas

Inicialmente é necessario fazer a conexão com a IA pq eu n vo escrever todos os peixes nao

````python
from google import genai
from google.genai import types
from config import chave, modelo
import os
import time


client = genai.Client(api_key=chave)



historico_peixes = ""

for i in range(5):
    prompt_text = f"""
    Crie uma lista de 15 peixes assimilados (mutantes) para um jogo RPG.
    Retorne APENAS os dados em formato CSV, separados por ponto e vírgula (;).
    NÃO use blocos de código markdown (sem ```), apenas o texto cru.

    Cabeçalho (incluir apenas se for o primeiro arquivo): Nome;Tamanho;Horario;Alimentacao;Agressividade;Caracteristicas

    IMPORTANTE: Não repita nomes ou características destes peixes já criados anteriormente:
    {historico_peixes}
    """

    try:
        response = client.models.generate_content(
            model=modelo,
            contents=prompt_text,
            config=types.GenerateContentConfig(
                temperature=0.7
            )
        )

        dados_csv = response.text.replace("```csv", "").replace("```", "").strip()

        nome_arquivo = f'peixe_lote_{i+1}.csv'
        with open(nome_arquivo, 'w', encoding='utf-8') as f:
            f.write(dados_csv)


        historico_peixes += "\n" + dados_csv

        time.sleep(2)

    except Exception as e:
        print(f" Erro no lote {i+1}: {e}")


print("\nProcesso finalizado.")
````
